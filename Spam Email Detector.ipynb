```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

data = {
    "message": [
        "Congratulations! You won a free lottery prize",
        "You have won $1000. Claim your prize now",
        "Free entry in a contest, click here",
        "Win a free mobile phone today",
        "You are selected for a cash reward",
        "Get free money by clicking this link",
        "Urgent! Claim your free gift now",
        "You have won a lucky draw",
        
        "Hi, how are you doing today?",
        "Please submit your assignment tomorrow",
        "Can we meet at 5 PM today?",
        "Your meeting is scheduled for Monday",
        "Please call me when you are free",
        "Don't forget to attend the class",
        "Your project presentation is tomorrow",
        "Let's have lunch together"
    ],

    "label": [
        "spam",
        "spam",
        "spam",
        "spam",
        "spam",
        "spam",
        "spam",
        "spam",

        "ham",
        "ham",
        "ham",
        "ham",
        "ham",
        "ham",
        "ham",
        "ham"
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

X = df["message"]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

vectorizer = TfidfVectorizer()

X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

model = MultinomialNB()

model.fit(X_train_vectorized, y_train)

# -----------------------------------------
# 6. Test model
# -----------------------------------------

y_pred = model.predict(X_test_vectorized)

accuracy = accuracy_score(y_test, y_pred)

print("\nModel Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

# -----------------------------------------
# 7. Predict new messages
# -----------------------------------------

print("\n--------------------------------")
print("SPAM EMAIL DETECTOR")
print("--------------------------------")

while True:

    message = input("\nEnter an email/message (or type 'exit' to stop): ")

    if message.lower() == "exit":
        print("Program stopped.")
        break

    message_vectorized = vectorizer.transform([message])

    prediction = model.predict(message_vectorized)[0]

    if prediction == "spam":
        print("Prediction: SPAM")
    else:
        print("Prediction: NOT SPAM")
```
